```python
import json
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def save_best_params(log_path: str, param_file_name: str):
    with open(log_path, 'r') as file:
        log = json.load(file)  # `data` will be a dictionary if the JSON structure is key-value pairs
    
    min_item = min(log , key=lambda x: x['value'])
    min_params = min_item['params']
    min_params = np.array(min_params)

    np.savetxt(param_file_name, min_params)
    return min_item['value']
```

```python
def graph_min(log_path: str):
    
    with open(log_path, 'r') as file:
        log = json.load(file)  # `data` will be a dictionary if the JSON structure is key-value pairs 
        
    values = [item['value'] for item in log]
    iterations = list(range(len(values)))  # Create a list of iteration numbers
    min_index = values.index(min(values))
    min_value = values[min_index] 
    
    sns.set_theme(rc={'figure.figsize':(15, 10)})
    sns.set_style("white")
    
    sns.lineplot(x = iterations, y = values, color="black")
    plt.scatter(min_index, min_value, color='red', s=100, label='Min Point') 
    plt.xlabel('Func Evaluation')
    plt.ylabel('RMSE')
    plt.grid(True)
    plt.show()
```
